```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline

sns.set_style('whitegrid')
titanic = sns.load_dataset('titanic')
titanic.head(10)
```

```python
print(titanic["embarked"].unique())
print(titanic["embarked"].isna().sum())
```

```python
# 1. Countplot: Passenger Count by Embarked Port
# ● Create a countplot to display the number of passengers who boarded from each
# port (embarked), separated by sex.
```

```python
sns.countplot(x='embarked',data=titanic,hue='sex',palette=['red', 'blue'])
plt.show()
```

```python
#  Observation 1 — Embarked distribution
# Most passengers embarked from Southampton (S).
# The majority of these passengers are male, showing a strong gender imbalance at this port.

# Observation 2 — Queenstown pattern
# Queenstown (Q) has the lowest number of passengers.
# The male–female counts are almost equal, suggesting many passengers may have been travelling as families.
```

```python
# Swarmplot: Age Distribution across Classes
# ● Create a swarmplot to visualize the distribution of age across passenger classes
# (pclass), categorized by sex.
```

```python
sns.swarmplot(x='class',y='age',data=titanic,hue='sex',palette=['red', 'blue'])
plt.legend(loc='upper right', bbox_to_anchor=(1.22, 0.98))
plt.show()
```

```python
# First Class — Age vs Sex
# Passengers below age 40 in First Class show a balanced distribution of males and females.
# For passengers above age 40, the majority are male, indicating a gender shift in older First‑Class travellers.

# Second Class — Age vs Sex
# Second Class shows almost equal numbers of male and female passengers overall.
# Children below age 10 are relatively few, suggesting fewer young families travelled in this class.

# Third Class — Age vs Sex
# Most Third‑Class passengers fall between ages 15 and 35, and the majority are male.
# There is a high number of children below age 10, indicating that many families travelled in Third Class.
```

```python
# Pie Chart: Gender Distribution
# ● Create a pie chart using Matplotlib to show the percentage of male vs. female
# passengers. - Add percentage labels using autopct - Use explode for one slice to
# highlight it.
```

```python
plt.pie(titanic['sex'].value_counts(), autopct='%1.1f%%', explode=[0, 0.1])
plt.show()
```

```python
#Most of the passangers are Male
```

```python
# Heatmap: Correlation Matrix
# ● Create a heatmap to show correlations between age, fare, pclass, sibsp, parch,
# and survived. (Use seaborn.heatmap with annotations)
```

```python
cor=titanic[['age', 'fare', 'pclass', 'sibsp', 'parch', 'survived']].corr()
sns.heatmap(cor, annot=True)
plt.show()
```

```python
#Passengers who paid higher fares had a better chance of survival.
```

```python
# Violin Plot: Fare by Class and Gender
# ● Create a violin plot to visualize fare distribution across classes, separated by
# gender.
```

```python

sns.violinplot(x='class',y='fare',data=titanic,hue='sex')
plt.show()
```

```python
# first class fare are very high
```

```python
# Scatter Plot: Age vs Fare
# ● Create a scatter plot to explore the relationship between age and fare, colored by
# survival status.
# ● Use a seaborn scatterplot with hue='survived'. Each plot must have titles, labels,
# and legends where needed.
```

```python
sns.scatterplot(x='age',y='fare',data=titanic,hue='survived')
plt.title('Titanic: Fare vs Age by Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```
