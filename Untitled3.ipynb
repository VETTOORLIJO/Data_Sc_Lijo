```python
from sklearn.metrics import roc_curve, auc
naive_bayes_model= GaussianNB()
naive_bayes_model.fit(X_train_scale,y_train)
naive_bayes_model_pred = naive_bayes_model.predict(X_test_scale)
naive_bayes_model_pred_accuracy =accuracy_score(y_test,naive_bayes_model_pred)
print("naive_bayes_model_pred_accuracy",naive_bayes_model_pred_accuracy)
print("naive_bayes_model_Classification  report \n",classification_report(y_test,naive_bayes_model_pred))
print("confusion_matrix\n",confusion_matrix(y_test,naive_bayes_model_pred))


print("roc")
fpr, tpr, thresholds = roc_curve(y_test, naive_bayes_model_pred)
roc_auc = auc(fpr, tpr)
plt.plot(fpr, tpr, label=f'AUC = {roc_auc:.3f}')
plt.plot([0,1], [0,1], 'r--')
plt.legend()
plt.show()
```

```python
from sklearn.naive_bayes import BernoulliNB

bernoulli_model = BernoulliNB()

bernoulli_model.fit(X_train_scale, y_train)

y_pred = bernoulli_model.predict(X_test_scale)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report\n")
print(classification_report(y_test, y_pred))
print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, y_pred))

# ROC
y_prob = bernoulli_model.predict_proba(X_test_scale)[:,1]

fpr, tpr, thresholds = roc_curve(y_test, y_prob)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(6,4))
plt.plot(fpr, tpr, label=f'AUC={roc_auc:.3f}')
plt.plot([0,1],[0,1],'r--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("BernoulliNB ROC Curve")
plt.legend()
plt.show()
```
